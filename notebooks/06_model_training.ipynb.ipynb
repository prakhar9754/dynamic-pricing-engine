#  Model Building

##  Objective

In this notebook, we build and evaluate two separate machine learning models:

1. **Classification Model**
   - Predict whether a user will purchase a product (`is_purchase`)

2. **Regression Model**
   - Predict optimal product price (`log_price`)

##  Important Note:

Both models are built independently using different targets and feature sets to avoid data leakage and ensure proper learning.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

# Classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Regression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

#saving model
import joblib
```

```python
df = pd.read_csv(r"C:\Users\hp\Desktop\PROJECTS\Dynamic_Pricing_Engine\data\final\final_featured_data.csv")

df.head()
```

##  Classification Model: Purchase Prediction

###  Goal:
Predict whether a user will make a purchase (`is_purchase`)

###  Features:
Behavioral and pricing features

###  Target:
`is_purchase`

```python
X_class = df.drop(columns=['is_purchase', 'price'])
y_class = df['is_purchase']
```

```python
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42
)
```

```python
clf = LogisticRegression(max_iter=1000)

clf.fit(X_train_c, y_train_c)
```

```python
y_pred_c = clf.predict(X_test_c)
```

```python
print("Accuracy:", accuracy_score(y_test_c, y_pred_c))
print("\nClassification Report:\n", classification_report(y_test_c, y_pred_c))
print("\nConfusion Matrix:\n", confusion_matrix(y_test_c, y_pred_c))
```

##  Classification Model Insights

- The model predicts purchase behavior based on user activity and pricing patterns
- Helps identify high-intent users
- Useful for demand estimation in dynamic pricing systems

##  Regression Model: Price Prediction

###  Goal:
Predict optimal product price (`log_price`)

###  Features:
User behavior and product demand features

###  Target:
`log_price`

```python
X_reg = df.drop(columns=['price', 'log_price'])
y_reg = df['log_price']
```

```python
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)
```

```python
reg = RandomForestRegressor(n_estimators=100, random_state=42)

reg.fit(X_train_r, y_train_r)
```

```python
y_pred_r = reg.predict(X_test_r)
```

```python
mae = mean_absolute_error(y_test_r, y_pred_r)
mse = mean_squared_error(y_test_r, y_pred_r)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_r, y_pred_r)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

##  Regression Model Insights

- The model predicts optimal pricing based on demand and behavior
- Helps in setting dynamic pricing strategies
- Can be used to maximize revenue and conversions

#  Conclusion

- Built two separate models:
  - Classification → predicts purchase behavior
  - Regression → predicts optimal price

- Ensured no data leakage by separating features properly

- These models together form the foundation of a Dynamic Pricing Engine
