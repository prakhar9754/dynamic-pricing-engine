# 📊 Exploratory Data Analysis (EDA)

## 🎯 Objective
In this notebook, we perform detailed Exploratory Data Analysis (EDA) on the eCommerce dataset to uncover patterns, trends, and insights that will help in building a Dynamic Pricing Engine.

## 📂 Dataset Overview
The dataset contains user interactions such as:
- Product views
- Add to cart
- Purchases

## 🔍 Key Goals:
- Understand customer behavior
- Analyze product pricing patterns
- Identify trends in sales and demand
- Detect anomalies and outliers

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\hp\Desktop\PROJECTS\Dynamic_Pricing_Engine\data\processed\ecommerce_cleaned.csv")
df.head()
```

## univariate analysis

##  Event Type Distribution

```python
sns.countplot(data=df, x='event_type')
plt.title("Event Type Distribution")
plt.show()
```

##  Price Distribution

```python
plt.figure(figsize=(8,5))  
sns.histplot(df['price'], bins=50, kde=True)
plt.title("Price Distribution")
plt.show()
```

##  Observation: Right Skewness in Price Distribution

The price distribution is **right-skewed (positively skewed)**, meaning:

- Most products have **lower prices**
- A few products have **very high prices**, creating a long tail on the right side

##  Interpretation

- The dataset contains a large number of **affordable products**
- High-priced items are **rare but significant**
- This imbalance can affect model performance

## ⚠️ Why this matters?

- Machine Learning models may get biased toward lower prices
- High variance in price can reduce prediction accuracy

## 💡 Solution

To handle skewness, we apply **log transformation**:

- Reduces extreme values impact
- Makes distribution more normal
- Improves model performance

##  Log Transformed Price Distribution

```python
df['log_price'] = np.log1p(df['price'])

sns.histplot(df['log_price'], bins=50, kde=True)
plt.title("Log Price Distribution")
plt.show()
```

##  Note on Log Transformation

The `log_price` feature is created temporarily for analysis purposes to better understand the distribution.

This transformation will be properly handled in the **Feature Engineering** stage before modeling.

## Bivariate Analysis

##  Price vs Event Type

```python
sns.boxplot(data=df, x='event_type', y='price')
plt.title("Price vs Event Type")
plt.show()
```

##  Top 10 Most Viewed Products

```python
top_products = df['product_id'].value_counts().head(10)

top_products.plot(kind='bar')
plt.title("Top 10 Products")
plt.show()
```

Time-based features such as hour and day were explored.

Therefore, time-related features are excluded from further analysis and modeling.
