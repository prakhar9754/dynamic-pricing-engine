#  Model Evaluation

##  Objective

This notebook performs detailed evaluation of the trained models:

- Classification Model → Predict purchase behavior
- Regression Model → Predict optimal pricing

##  Note

Basic evaluation was performed during training.
This notebook focuses on **final performance analysis and interpretation**.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

# Classification metrics
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)

# Regression metrics
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score
)
```

```python
df = pd.read_csv(r"C:\Users\hp\Desktop\PROJECTS\Dynamic_Pricing_Engine\data\final\final_featured_data.csv")
```

```python
# Classification
X_class = df.drop(columns=['is_purchase', 'price'])
y_class = df['is_purchase']

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42
)

# Regression
X_reg = df.drop(columns=['price', 'log_price'])
y_reg = df['log_price']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)
```

##  Classification Model Evaluation

We evaluate the model using:

- Precision → correctness of predicted purchases  
- Recall → ability to capture actual purchases  
- F1 Score → balance between precision and recall  
- Confusion Matrix → detailed prediction breakdown  

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression(max_iter=1000)
clf.fit(X_train_c, y_train_c)
```

```python
y_pred_c = clf.predict(X_test_c)
```

```python
print("Precision:", precision_score(y_test_c, y_pred_c))
print("Recall:", recall_score(y_test_c, y_pred_c))
print("F1 Score:", f1_score(y_test_c, y_pred_c))

print("\nClassification Report:\n", classification_report(y_test_c, y_pred_c))
print("\nConfusion Matrix:\n", confusion_matrix(y_test_c, y_pred_c))
```

##  Insights

- High precision → fewer false purchase predictions
- High recall → captures more real buyers
- F1 score balances both

This model helps identify high-intent users for pricing strategies.

##  Regression Model Evaluation

We evaluate the model using:

- MAE → average error in prediction  
- RMSE → penalizes large errors  
- R² Score → overall model fit  

```python
from sklearn.ensemble import RandomForestRegressor

reg = RandomForestRegressor(n_estimators=100, random_state=42)
reg.fit(X_train_r, y_train_r)
```

```python
y_pred_r = reg.predict(X_test_r)
```

```python
mae = mean_absolute_error(y_test_r, y_pred_r)
mse = mean_squared_error(y_test_r, y_pred_r)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_r, y_pred_r)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
y_pred_price = np.expm1(y_pred_r)  #actual
y_actual_price = np.expm1(y_test_r) #actual 
```

##  Insights

- MAE shows average pricing error
- RMSE highlights large prediction mistakes
- R² shows how well model explains price variation

Model can be used for dynamic pricing decisions.

- Classification model predicts purchase likelihood
- Regression model estimates optimal price

Both models together enable a data-driven dynamic pricing system.
