# Data Cleaning - Dynamic Pricing Engine

## Objective
To clean and preprocess the dataset by handling missing values, duplicates, incorrect data types, and inconsistencies.

## Input
ecommerce_sample.csv

## Output
Cleaned dataset ready for exploratory data analysis

```python
import pandas as pd
import numpy as np
```

## Load Dataset

```python
df = pd.read_csv(r"C:\Users\hp\Desktop\PROJECTS\Dynamic_Pricing_Engine\data\raw\ecommerce_sample.csv")
df.head()
```

```python
df.info()
```

## Remove Duplicate Records

Duplicate records can bias analysis and should be removed.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

## Handle Missing Values

Missing values in critical columns such as user_id and product_id can affect downstream analysis.

```python
df.isnull().mean() * 100
```

```python
df['brand'] = df['brand'].fillna('unknown')
```

```python
df['category_code'] = df['category_code'].fillna('unknown')
```

## Handling Missing Values in Categorical Features

The dataset contains missing values in the following columns:

- category_code (~32.6%)
- brand (~13.9%)

The `brand` column has relatively low missing values and is retained by imputing with a placeholder value 'unknown'.

The `category_code` column has moderate missing values. However, since product category is an important feature for understanding pricing behavior, the column is retained and missing values are replaced with 'unknown'.

This approach ensures:
- No significant data loss
- Preservation of important categorical information
- Compatibility with downstream encoding and modeling

## Checking value cases

```python
df['event_type'].value_counts()
```

## Convert Data Types

Ensure appropriate data types for analysis.

```python
df['event_time'] = pd.to_datetime(df['event_time'], errors='coerce')
```

## Remove Invalid Records

Drop rows where datetime conversion failed.

```python
df = df.dropna(subset=['event_time'])
```

## Price Cleaning

Remove invalid and unrealistic values from the price column.

```python
df['price'].describe()
```

## Handle Outliers

Remove extreme price values using the IQR method.

```python
q1 = df['price'].quantile(0.25)
q3 = df['price'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df = df[(df['price'] >= lower_bound) & (df['price'] <= upper_bound)]
```

## Standardize Categorical Values

Ensure consistency in categorical columns such as event_type.

```python
df['event_type'] = df['event_type'].str.lower().str.strip()
```

## Final Data Validation

Verify that the dataset is clean and consistent.

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

## Saving Clean Dataset

```python
df.to_csv(r"C:\Users\hp\Desktop\PROJECTS\Dynamic_Pricing_Engine\data\processed\ecommerce_cleaned.csv", index=False)
```
